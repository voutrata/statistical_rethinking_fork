```python
import numpy as np
```

```python
p = 0.7
l = np.array([p, 1-p])
-sum(l*np.log2(l))
```

```python
l = np.array([1/3, 1/3, 1/3])
-sum(l*np.log(l))
```

```python
def compute_entropy(l):
    return -sum(l*np.log2(l))

def compute_divergence(l1, l2):
    return sum(l2*np.log2(l2/l1))
```

```python
l1 = np.array([0.2] * 5)
l2 = np.array([0.8, 0.1, 0.05, 0.025, 0.025])
l3 = np.array([0.05, 0.15, 0.7, 0.05, 0.05])
```

```python
print(compute_entropy(l1))
print(compute_entropy(l2))
print(compute_entropy(l3))
```

```python
print("Island 1 approximating othes")
print(f"\tIsland2: {compute_divergence(l1, l2)}")
print(f"\tIsland3: {compute_divergence(l1, l3)}")
print("Island 2 approximating othes")
print(f"\tIsland1: {compute_divergence(l2, l1)}")
print(f"\tIsland3: {compute_divergence(l2, l3)}")
print("Island 3 approximating othes")
print(f"\tIsland1: {compute_divergence(l3, l1)}")
print(f"\tIsland2: {compute_divergence(l3, l2)}")
```
